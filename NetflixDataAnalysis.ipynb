```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
df = pd.read_csv('netflix_titles.csv', encoding='latin-1')
```

```python
print(df.info())
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
print(df.info())
```

```python
df.head(2)
```

```python
# Filling Missing Values
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Unknown')
```

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
# Feature Engineering
df["month"] = df["date_added"].dt.month
df["year"] = df["date_added"].dt.year
df[["total_time","duration_type"]] = df["duration"].str.split(" ", expand = True)
```

```python
sns.countplot(data = df, x = 'type', palette = 'pastel', hue = "type")
plt.title("Majority between Movies and TV Shows")
plt.show()
```

```python
counts = df["year"].value_counts().sort_index()
sns.lineplot(x=counts.index, y=counts.values, marker='o')
plt.title("Content added over Years")
plt.xlabel("Year")
plt.ylabel("Number of Content Added")
plt.show()
```

```python
# country may contain several values so split
df['country'] = df['country'].str.split(',')
df = df.explode('country')
df['country'] = df['country'].str.strip()
```

```python
df = df.reset_index(drop=True)
top_countries = df['country'].value_counts().head(10)
sns.countplot(y='country', data=df, order=top_countries.index, palette = "pastel", hue = "country")
plt.title("Top 10 Countries with the Most Content")
plt.xlabel("Number of Content's")
plt.ylabel("Country")
plt.show()
```

```python
sns.countplot(y='rating', data=df, palette = "pastel", hue = "rating")
plt.title("Most Common Ratings")
plt.xlabel("Count of ratings")
plt.ylabel("Types of Ratings")
plt.show()
```

```python
year_type_counts = df.groupby(['year', 'type']).size().reset_index(name='count')

plt.figure(figsize=(10, 5))
sns.lineplot(data=year_type_counts,x='year',y='count',hue='type',marker='o')

plt.xticks(rotation=45)
plt.title("Trend of Movies vs TV Shows Over Years")
plt.show()
```

```python
df.rename(columns={'listed_in': 'genre'}, inplace=True)
# Separating geners
df['genre'] = df['genre'].str.split(',')
df = df.explode('genre')
df['genre'] = df['genre'].str.strip()
df = df.reset_index(drop=True)
```

```python
top_genres = df['genre'].value_counts().head(10)

sns.barplot(y=top_genres.index,x=top_genres.values, palette = "pastel", hue = top_genres.values)
plt.title("Top 10 Most Popular Genres")
plt.xlabel("Popularity Count")
plt.ylabel("Genre")
plt.show()

```

```python
df.to_csv("cleaned_netflix.csv", index=False)
```
